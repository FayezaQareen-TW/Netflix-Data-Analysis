```python
import pandas as pd

# 1. Unzip the file you uploaded directly inside Google Colab
!unzip -o netflix_titles.csv.zip

# 2. Now load the extracted CSV file
df = pd.read_csv("netflix_titles.csv")

# 3. Print success confirmation for your portfolio
print("\n🎉 Success! Dataset extracted and loaded.")
print(f"Data Dimensions: {df.shape[0]} Rows and {df.shape[1]} Columns\n")

# Show the first 5 rows of data
df.head()
```

```python
# Check for missing data in each column
print("--- Missing Values Per Column ---")
df.isnull().sum()
```

```python
import matplotlib.pyplot as plt
import seaborn as sns

# 1. Set a clean, modern chart background style
sns.set_theme(style="whitegrid")

# 2. Create a bar chart comparing Movies vs TV Shows
plt.figure(figsize=(7, 4.5))
sns.countplot(data=df, x='type', palette='Reds_r')

# 3. Add professional titles and clean text labels
plt.title('Netflix Content Strategy: Movies vs TV Shows', fontsize=14, fontweight='bold', pad=15)
plt.xlabel('Content Category', fontsize=12)
plt.ylabel('Total Count on Platform', fontsize=12)

# 4. Display the visual chart
plt.show()
```

```python
import matplotlib.pyplot as plt
import seaborn as sns

# 1. Get the top 5 countries producing content
top_countries = df['country'].value_counts().head(5)

# 2. Set the chart background style
sns.set_theme(style="whitegrid")
plt.figure(figsize=(8, 5))

# 3. Create a horizontal bar chart
sns.barplot(x=top_countries.values, y=top_countries.index, hue=top_countries.index, palette='Reds_r', legend=False)

# 4. Add clear titles and labels
plt.title('Top 5 Content Producing Countries on Netflix', fontsize=14, fontweight='bold', pad=15)
plt.xlabel('Total Number of Titles', fontsize=12)
plt.ylabel('Country', fontsize=12)

plt.show()

# 5. Print the raw counts below the chart
print("--- Exact Production Counts ---")
print(top_countries)
```

```python
import matplotlib.pyplot as plt
import seaborn as sns

# 1. Get the top 5 most common age ratings on Netflix
top_ratings = df['rating'].value_counts().head(5)

# 2. Set the chart background style
sns.set_theme(style="whitegrid")
plt.figure(figsize=(7, 4.5))

# 3. Create a clean vertical bar chart
sns.barplot(x=top_ratings.index, y=top_ratings.values, hue=top_ratings.index, palette='Reds_r', legend=False)

# 4. Add clear titles and labels
plt.title('Top 5 Content Age Ratings on Netflix', fontsize=14, fontweight='bold', pad=15)
plt.xlabel('Age Rating Category', fontsize=12)
plt.ylabel('Total Number of Titles', fontsize=12)

plt.show()
```
